```c++
#include <iostream>
#include "includeLibraries.h"
#include <opencv2/core.hpp>
#include <opencv2/cudaarithm.hpp>
#include <opencv2/cudaimgproc.hpp>
#include <opencv2/cudawarping.hpp>
```

```c++
std::cout << "OpenCV version: " << CV_VERSION << std::endl;

int gpu_count = cv::cuda::getCudaEnabledDeviceCount();
std::cout << "CUDA devices: " << gpu_count << std::endl;
```

```c++
if (gpu_count > 0) {
    cv::cuda::DeviceInfo info(0);

    std::cout << "Device name: " << info.name() << std::endl;
    std::cout << "Compute capability: "
                << info.majorVersion() << "."
                << info.minorVersion() << std::endl;

    std::cout << "Total memory: "
                << info.totalMemory() / (1024 * 1024) << " MB" << std::endl;

    cv::cuda::setDevice(0);

    cv::Mat cpu = cv::Mat::ones(512, 512, CV_8UC1) * 255;
    cv::cuda::GpuMat gpu;

    gpu.upload(cpu);

    cv::cuda::GpuMat out;
    cv::cuda::threshold(gpu, out, 128, 255, cv::THRESH_BINARY);

    cv::Mat result;
    out.download(result);

    std::cout << "CUDA threshold executed successfully" << std::endl;
} else {
    std::cout << "CUDA not available in OpenCV build" << std::endl;
}
```

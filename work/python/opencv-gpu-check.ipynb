```python
import cv2

# Check if OpenCV is built with CUDA support
print("OpenCV version:", cv2.__version__)
print("CUDA enabled:", cv2.cuda.getCudaEnabledDeviceCount() > 0)

if cv2.cuda.getCudaEnabledDeviceCount() > 0:
    print("Number of CUDA devices:", cv2.cuda.getCudaEnabledDeviceCount())
    for i in range(cv2.cuda.getCudaEnabledDeviceCount()):
        cv2.cuda.setDevice(i)
        print(f"Device {i}: {cv2.cuda.getDevice()}")
else:
    print("No CUDA-enabled devices found. OpenCV GPU support is not available.")
```
